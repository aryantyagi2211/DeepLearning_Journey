## Step 11: Build the deep neural network

````python
model = keras.Sequential([
    keras.Input(shape=(28, 28)),
    layers.Flatten(),
    layers.Dense(128, activation="relu"),
    layers.Dense(64, activation="relu"),
    layers.Dropout(0.2),
    layers.Dense(10, activation="softmax")
])
````

### Complete architecture

The model performs the following transformation:

**28 × 28 image → 784 values → 128 neurons → 64 neurons → 10 output probabilities**

### Line-by-line explanation

```python
model = keras.Sequential([
```

- Creates a Sequential neural network.
- Sequential means that layers are arranged one after another.

```python
keras.Input(shape=(28, 28)),
```

- Defines the expected input shape.
- Each input is one 28 × 28 image.
- The batch dimension is handled automatically.

```python
layers.Flatten(),
```

- Converts the 28 × 28 matrix into a one-dimensional vector.
- Number of values: `28 × 28 = 784`.
- Flatten changes only the shape; it does not learn weights.

```python
layers.Dense(128, activation="relu"),
```

- Creates the first hidden layer with 128 neurons.
- `Dense` means every neuron receives input from every value in the previous layer.
- `relu` keeps positive values and converts negative values to zero.

```python
layers.Dense(64, activation="relu"),
```

- Creates the second hidden layer with 64 neurons.
- This makes the network deeper and allows it to learn more complex combinations of features.

```python
layers.Dropout(0.2),
```

- Randomly turns off 20% of the neurons during each training step.
- This helps reduce overfitting.
- Dropout is disabled during testing and prediction.

```python
layers.Dense(10, activation="softmax")
```

- Creates the output layer.
- There are 10 neurons because there are 10 possible digits.
- Softmax converts the output values into probabilities.
- The probabilities add up to approximately 1.

### Inference

The hidden layers learn patterns such as edges, curves, loops, and combinations of strokes. The output layer estimates which digit is most likely.

### Understanding parameter calculation

For a Dense layer:

**Parameters = input values × neurons + biases**

First Dense layer:

- Inputs = 784
- Neurons = 128
- Parameters = `784 × 128 + 128 = 100,480`

Second Dense layer:

- Inputs = 128
- Neurons = 64
- Parameters = `128 × 64 + 64 = 8,256`

Output layer:

- Inputs = 64
- Neurons = 10
- Parameters = `64 × 10 + 10 = 650`

### Inference

Weights and biases are the values learned by the neural network during training. Flatten and Dropout do not have trainable parameters.

## Step 26: Generate a classification report

````python
report = classification_report(
    y_test,
    predicted_classes
)

print(report)
````

### Explanation of the report

**Precision**

Of all images predicted as a particular digit, how many predictions were correct?

**Recall**

Of all actual images belonging to a particular digit, how many did the model identify correctly?

**F1-score**

A balanced combination of precision and recall.

**Support**

The number of actual test images belonging to each digit.

### Inference

Values closer to 1 indicate better performance. The report helps us check whether the model performs well across all digits rather than only looking at overall accuracy.

## Classroom summary

> A neural network learns handwritten digits by converting each image into pixel values. During training, it compares its predictions with the correct labels, calculates the error, and adjusts its weights. After repeated training, it learns patterns such as lines, curves, and loops that help distinguish digits from 0 to 9.
